### GroupBy_AggregateFunctions

```python
from pyspark.sql import SparkSession
```

```python
spark= SparkSession.builder.appName("DataFrame").getOrCreate()
```

```python
spark
```

```python
df_pyspark= spark.read.csv("./datasets/test3.csv", header= True, inferSchema= True)
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```

#### gropuBy and aggrigate functions work together

```python
df_pyspark.groupBy("Name").sum().show()
```

```python
#gropby Departments

df_pyspark.groupBy("Departments").sum().show()
```

```python
#mean
df_pyspark.groupBy("Departments").mean().show()
```

```python
#how many employee are working 

df_pyspark.groupBy("Departments").count().show()
```

```python
#applying aggrigate functions directly

df_pyspark.agg({'Salary':'sum'}).show()
```

```python
#Find out who is getting more salary

df_pyspark.groupBy("Name").max().show()
```

```python
#depertment 

df_pyspark.groupBy("Departments").max().show()
```
