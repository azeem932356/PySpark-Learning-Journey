#### 1. Dropping Columns

```python
#Creating SparkSession

from pyspark.sql import SparkSession
spark= SparkSession.builder.appName("Practise").getOrCreate()
```

```python
df_pyspark= spark.read.csv("./datasets/test2.csv", header= True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
#Dropping columns

df_pyspark.drop('Name', 'age').show()
```

```python
df_pyspark.drop('Name').show()
```

```python
#Droping NA values

df_pyspark.na.drop().show()    #You can see where ever a NA values are there it dropped entire row
```

#### drop()   --> has 3 options drop(how:   , thresh:    , subset:    )

#### drop(how:) --> has 2 option "any", "all"

```python
#lets see "all"  first

df_pyspark.na.drop(how=  "all").show()  #If how= "all", drop a row only if all its values are null (NA)
```

```python
#Lets see "any"

df_pyspark.na.drop(how= "any").show()    #If how= "any" , drop a row if it contains any null (NA) values
```

#### threshold

```python
#thresh

df_pyspark.show()
```

```python
df_pyspark.na.drop(how= "any", thresh= 2).show()   ##thresh= 2  --> means keep a row if it has at least 2 non-null values, (You can look above df_pyspark last row has only one non-null value so it drops)
```

#### subset

```python
# subset tells spark which column it should check for null values

df_pyspark.na.drop(subset= ["Experience"]).show()
```
