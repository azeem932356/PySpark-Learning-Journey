#### 1. Dropping Columns

```python
#Creating SparkSession

from pyspark.sql import SparkSession
spark= SparkSession.builder.appName("Practise").getOrCreate()
```

```python
df_pyspark= spark.read.csv("./datasets/test2.csv", header= True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
#Dropping columns

df_pyspark.drop('Name', 'age').show()
```

```python
df_pyspark.drop('Name').show()
```

```python
#Droping NA values

df_pyspark.na.drop().show()    #You can see where ever a NA values are there it dropped entire row
```

#### drop()   --> has 3 options drop(how:   , thresh:    , subset:    )

#### drop(how:) --> has 2 option "any", "all"

```python
#lets see "all"  first

df_pyspark.na.drop(how=  "all").show()  #If how= "all", drop a row only if all its values are null (NA)
```

```python
#Lets see "any"

df_pyspark.na.drop(how= "any").show()    #If how= "any" , drop a row if it contains any null (NA) values
```

#### threshold

```python
#thresh

df_pyspark.show()
```

```python
df_pyspark.na.drop(how= "any", thresh= 2).show()   ##thresh= 2  --> means keep a row if it has at least 2 non-null values, (You can look above df_pyspark last row has only one non-null value so it drops)
```

#### subset

```python
# subset tells spark which column it should check for null values

df_pyspark.na.drop(subset= ["Experience"]).show()
```

```python
#You can't all three at a time

df_pyspark.na.drop(how= 'any', subset= ['Name']).show()
```

```python
df_pyspark.show()
```

```python
df_pyspark.na.drop(thresh= 1, subset= ['Name']).show()
```

#### Filling the missing values

#### df_pyspark.na.fill(takes two arguments 1. value, 2. thresh)

```python
df_pyspark.show()
```

```python
df_pyspark.na.fill('Missing Values').show()
```

```python
#For non string values

df_pyspark.na.fill(0).show()
```

```python
#For entire data

df_pyspark.na.fill({
    'Name': 'Missing value', 
    'age': 0, 
    'Experience': 0, 
    'Salary': 0
}).show()
```

```python
df_pyspark.printSchema()
```

```python
#For specific column {Those are string values}

df_pyspark.na.fill("Missing values", subset=["Name"]).show()
```

```python
#For Non string values
df_pyspark.na.fill(0, subset= ['age']).show()
```

```python
#Replacing with Mean, Median and Mode values

from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=["age", 'Experience', 'Salary'],
    outputCols=["{}_impted".format(c) for c in ['age', 'Experience', 'Salary']]
).setStrategy('mean')
```

```python
#add imputation also to df

imputer.fit(df_pyspark).transform(df_pyspark).show()
```

```python
from pyspark.sql import functions as F

# Calculate average and rename the resulting column
df_pyspark.select(F.avg("Salary").alias("average_salary")).show()
```

```python
##Mode

from pyspark.ml.feature import Imputer

imputed= Imputer(
    inputCols= ["age", 'Experience', 'Salary'],
    outputCols= ["{}_Imputed". format(c) for c in ["age", 'Experience', 'Salary']]
).setStrategy("Mode")
```

```python
#mode
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

```python
#Median

from pyspark.ml.feature import Imputer

imputed= Imputer(
    inputCols= ["age", 'Experience', 'Salary'],
    outputCols= ["{}_Imputed". format(c) for c in ["age", 'Experience', 'Salary']]
).setStrategy("Median")
```

```python
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
