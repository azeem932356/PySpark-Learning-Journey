```python
from pyspark.sql import SparkSession
```

```python
spark= SparkSession.builder.appName("dataframe").getOrCreate()
```

```python
spark
```

```python
df_pyspark= spark.read.csv("./datasets/test1.csv", header= True, inferSchema= True)
```

```python
df_pyspark.show(3)
```

### Filter Operations


```python
#Salary of the people less than or equal to 20000

df_pyspark.filter("Salary <= 20000").show()
```

```python
df_pyspark.filter(df_pyspark["Salary"]<= 20000).show()
```

```python
#selecting some columns only

df_pyspark.filter("Salary <= 20000").select(["Name", "age"]).show()
```

```python
#Another way of filter

df_pyspark.filter(df_pyspark["Salary"]<= 20000).show()
```

```python
#using multiple conditions at a time

df_pyspark.filter("Salary<= 20000 and Salary <= 15000" ).show()
```

```python
# or (|) both are same

df_pyspark.filter("Salary<= 20000 or Salary <= 15000" ).show()
```

```python
df_pyspark.filter((df_pyspark["Salary"]<20000) | (df_pyspark["Salary"]>=15000)).show()
```

```python
df_pyspark.show()
```

```python
# ~ not  condition, basically what you get is other than your original condition

df_pyspark.filter(~(df_pyspark["Salary"]>=20000)).show()
```
