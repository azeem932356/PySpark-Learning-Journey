```python
#First create sparksession

from pyspark.sql import SparkSession
```

```python
spark= SparkSession.builder.appName("Dataframe").getOrCreate()
```

```python
spark
```

```python
#reading files

df_pyspark= spark.read.csv(
    "datasets/test1.csv",
    header= True, 
    inferSchema=True
)
```

```python
df_pyspark.show(4)
```
